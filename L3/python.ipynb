| Escape | Meaning |
|---|---|
| `\n` | new line |
| `\t` | tab space |
| `\\` | a real backslash |

````python
print("Vishavs\traj")
print("a\nb\\c\"\tok")
````

````python
print("12345678\t|")
print("Vishavs\t|")   # 7 chars -> 1 space to reach column 8
print('b\\c"\t|')     # 4 chars -> 4 spaces to reach column 8
print("a\t|")         # 1 char  -> 7 spaces
print("123456789\t|") # 9 chars -> jumps to column 16
````

# 3️⃣ Program 3 — Variables & Data Types
### ⏱️ 8–15 min &nbsp;|&nbsp; *(file: 5.py)*

A **variable** is a labelled box that stores a value so you can use it later. You put something in the box with `=` and use the box by its name.

**The four core data types beginners need:**

| Type | Name | Examples |
|---|---|---|
| `str` | String — text, always in quotes | `"Ravi"`, `'r'` |
| `int` | Integer — whole numbers | `21`, `-4`, `100` |
| `float` | Numbers with a decimal point | `5.9`, `1.5` |
| `bool` | Boolean — only two values | `True`, `False` |

Use `type()` to ask Python what kind of value something is.


````python
# types

name = "vishvas"
charcter = 'r'
Name = "vkp"
num = 1
yes = True
per = 1.5

print(name)
print(Name)

print(type(name))
print(type(charcter))
print(type(Name))
print(type(num))
print(type(yes))
print(type(per))
````

**Expected output**
```
vishvas
vkp
<class 'str'>
<class 'str'>
<class 'str'>
<class 'int'>
<class 'bool'>
<class 'float'>
```

> 🔎 **Notice:** `name` and `Name` are **two different boxes**. Python is case-sensitive.
> Also, Python has **no separate character type** — `'r'` is just a `str` of length 1. Single and double quotes both make strings.

> ⚠️ **Watch Out** — Variable names cannot start with a number or contain spaces. Use `first_name`, not `first name` or `1name`.

# 4️⃣ Program 4 — Printing Variables, `sep` and f-strings
### ⏱️ 15–19 min &nbsp;|&nbsp; *(file: 3.py)*

Now that you can store a value in a variable, here are three ways to print it:

1. `print(name)` → just the value
2. `print("Number:", number, sep="")` → `sep` controls what goes **between** items (default is one space)
3. `print(f"Number:{number}")` → an **f-string**: put a variable inside `{ }` right in the text


````python
name = "Vishvas"
number = "0"

print(name)

print("Number:", number)
#sep = separator
print("Number:", number, sep="")
print(f"Number:{number}")
````

**Expected output**
```
Vishvas

Number:0
```

> ⚠️ **Watch Out** — `print(name)` prints the **value** stored in the box. `print("name")` prints the **word** `name`. Quotes change everything!

# 8️⃣➕ Program 8B — Logical Operators (`and`, `or`, `not`)
### ⏱️ +4 min bonus &nbsp;|&nbsp; *(no file — type it live)*

Comparison operators answer **one** question. Logical operators let you **join two or more questions** together.

| Operator | Meaning | True when… |
|---|---|---|
| `and` | both must be true | *"IF it rains **and** I have an umbrella"* |
| `or` | at least one must be true | *"IF it is Saturday **or** Sunday"* |
| `not` | flips the answer | `not True` → `False` |

**Truth tables — say these aloud**

| A | B | `A and B` | `A or B` |
|---|---|---|---|
| True | True | True | True |
| True | False | False | True |
| False | True | False | True |
| False | False | False | False |


````python
age = 20
marks = 45

print(age >= 18 and marks >= 40)   # True  and True  -> True
print(age >= 18 and marks >= 50)   # True  and False -> False

print(age >= 60 or marks >= 50)    # False or False  -> False
print(age >= 60 or marks >= 40)    # False or True   -> True

print(not (age < 18))              # not False -> True
print(not (age >= 18))             # not True  -> False

print(age > 10 and age < 30)       # both sides written in full
print(10 < age < 15)               # Python also allows chaining
````

**Expected output**
```
True
False
False
True
True
False
True
False
```

> ⚠️ **Watch Out** — Write `age > 10 and age < 20`, **not** `age > 10 and < 20`. Each side of `and` / `or` must be a **complete** comparison.

# 9️⃣ Program 9 — Decisions: `if` / `elif` / `else`
### ⏱️ 40–50 min &nbsp;|&nbsp; *(file: 9.py)*

**Idea:** Programs choose different paths depending on a condition — just like real life: *"IF it rains, take an umbrella, ELSE wear sunglasses."*

> 🔎 **Indentation matters!** Python uses **indentation (spaces)** to know which lines belong inside the `if`. Use a consistent **4 spaces**. The line before an indented block **always ends with a colon `:`**.

````python
age = input("Enter your age")

age = int(age)

if age >= 18:
    print("You can watch this movie")
else:
    print("sorry")
````

**Sample run**
```

```

### Choosing between many options with `elif`
`elif` = "else if". Python checks the conditions **top to bottom** and runs the **first one that is True** — then skips the rest.

````python
marks = int(input("Enter marks: "))

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")
````

**Sample run**
```
Enter marks: 82
Grade B
```

> ⚠️ **Watch Out — indentation & colons**
> Forgetting the `:` or mixing tabs and spaces causes `IndentationError` or `SyntaxError`. Pick 4 spaces and stay consistent.

# 🔟 Program 10 — `for` Loop & `range()`
### ⏱️ 50–54 min &nbsp;|&nbsp; *(file: 10.py)*

**Idea:** A loop repeats instructions so you don't copy-paste.

- `range(5)` → `0, 1, 2, 3, 4` — **starts at 0, stops *before* 5**
- `range(2, 6)` → `2, 3, 4, 5` — start at 2, stop before 6
- `range(5, 0, -1)` → `5, 4, 3, 2, 1` — counting backwards

````python
for i in range(5):
    print(i)

for i in range(2, 6):
    print(i)
````

> ⚠️ **Watch Out — off-by-one:** `range(5)` gives **0–4**, *not* 1–5. To get 1–5 use `range(1, 6)`.

# 1️⃣1️⃣ Program 11 — `while` Loop
### ⏱️ 54–57 min &nbsp;|&nbsp; *(file: 11.py)*

A `while` loop repeats **as long as a condition stays True**.

Every `while` loop needs three things:
1. **Start** the counter → `i = 0`
2. **Check** the condition → `while i < 5:`
3. **Change** the counter inside the loop → `i = i + 1`

> ⚠️ **Watch Out — infinite loops:** if you forget step 3, the loop runs **forever**. In a notebook, press the ⏹️ **stop** button to interrupt it.

**Rule of thumb:** use `for` when you know *how many times*; use `while` when you loop *until something happens*.

````python
i = 0
while i < 5:
    print(i)
    i = i + 1
````

````python
# The same result written as a for loop
for i in range(5):
    print(i)
````

**Expected output (both cells)**
```
0
1
2
3
4
```

## ❌ Common Beginner Mistakes — Quick Reference

| ❌ Common Mistake | ✅ How to Fix It |
|---|---|
| Forgetting quotes: `print(Hello)` | `print("Hello")` |
| Using `=` instead of `==` | `==` to compare, `=` to assign |
| Maths on raw input: `age + 1` | Convert first: `int(input(...))` |
| Missing colon: `if x > 5` | `if x > 5:` |
| Wrong indentation inside `if`/loop | Use exactly 4 spaces, be consistent |
| Mismatched brackets: `print("Hi"` | Close every `(` and `"` you open |
| Half a condition: `x > 10 and < 20` | Write both sides: `x > 10 and x < 20` |
| Using `&&` / `\|\|` from other languages | Python uses `and` / `or` |
| Expecting `range(5)` to give 1–5 | It gives 0,1,2,3,4 — use `range(1,6)` |
| `while` loop never ends | Update the counter inside the loop |
| Case error: `Print` / `PRINT` | Python is case-sensitive: use `print` |

> 💡 **Teaching tip:** when an error appears, don't hide it — read the **last line** of the red error message aloud. It names the problem (`SyntaxError`, `NameError`, `TypeError`).
